### Import the Data
Let's extract the our **Titanic data** from the .csv file, create a  pandas DataFrame and look at the available indicators:

- ***Survived***: Outcome of survival (0 = No; 1 = Yes)
- ***Pclass***: Socio-economic class (1 = Upper class; 2 = Middle class; 3 = Lower class)
- ***Name***: Name of passenger
- ***Sex***: Sex of the passenger
- ***Age***: Age of the passenger (Some entries contain NaN)
- ***SibSp***: Number of siblings and spouses of the passenger aboard
- ***Parch***: Number of parents and children of the passenger aboard
- ***Ticket***: Ticket number of the passenger
- ***Fare***: Fare paid by the passenger
- ***Cabin***: Cabin number of the passenger (Some entries contain NaN)
- ***Embarked***: Port of embarkation of the passenger (C = Cherbourg; Q = Queenstown; S = Southampton)

```python
Q1 = df["Age"].quantile(0.25) 
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1 # formula
print (Q1)
print (Q3)
print(IQR) 
```

```python
lower_bound = Q1 - 1.5 * IQR 
upper_bound = Q3 + 1.5 * IQR
print(lower_bound) 
print(upper_bound)
```

```python
df_filtered = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)]

print(f"Original Rows: {df.shape[0]}, Filtered Rows: {df_filtered.shape[0]}")
```

```python
# Imputation function
def impute_age(cols): # make list required
    Age = cols[0] #  age then Pclass
    Pclass = cols[1]

    if pd.isnull(Age): # check first if age is null so fill it 
       
        if Pclass == 1:  # above we plot boxplot where we see the mid line or quantile or mean is 37 for Pclass 1 
            return 37

        elif Pclass == 2:  # same mean is 29 for Pclass 2
            return 29

        else: # same mean is 24 for Pclass 3
            return 24

    else: # else if age is not null return it without change or fill
        return Age
```

```python
df['Age']=df[['Age','Pclass']].apply(impute_age, axis =1)
```
